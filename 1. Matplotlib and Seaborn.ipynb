# Simple Plotting

```python
# Plotting with line : Line Plot

week=[1,2,3,4]
prc=[45,56,34,23]
pl.plot(week,prc)
pl.show
```

```python
# Plotting without line: Scatter Plot

import matplotlib.pyplot as pl
week=[1,2,3,4]
prc=[45,56,34,23]
pl.scatter(week,prc)
pl.ylabel("Price")
pl.xlabel("Week")
pl.title("Data")
pl.show()
```

```python
# Multiple Lines

import numpy as np
week=[11,22,33,44]
prc=[45,56,34,23]
y1 = np.array([3, 8, 1, 10])
y2 = np.array([6, 2, 7, 11])
pl.plot(week)
pl.plot(prc)
pl.show()
```

# Markers

```python
# Plotting with line : Line Plot

pl.figure(figsize=(7,5))
week=[1,2,3,4]
prc=[45,56,34,23]
pl.plot(week,prc,'r', linewidth=4, linestyle='-', marker='^', markersize=15, markeredgecolor='k')
pl.grid(True)
pl.title("Data")
pl.show
```

```python
# Plotting without line: Scatter Plot

import matplotlib.pyplot as pl
week=[1,2,3,4]
prc=[45,56,34,23]
pl.scatter(week,prc, marker='x', c=['k','b','y','m'], s=65)
pl.title("Data")
pl.show()
```

# Bar Chart

```python
# Single Bar

import numpy as np
week=[1,2,3,4]
prc=[45,56,34,23]
prc2=[48,58,53,89]
x=np.arange(4)
pl.xlim(-2,5)
pl.xticks([1,3,5])
pl.bar(week,prc, width=[.4,.4,.3,.5], color=['r','k','b','m'])
pl.ylabel("Price")
pl.xlabel("Week")
pl.show()
```

```python
# Multiple bar

import numpy as np
week=[1,2,3,4]
prc=[45,56,34,23]
prc2=[48,58,53,89]
x=np.arange(4)
pl.xlim(-1,5)
pl.bar(x+0.00,prc, width=0.25, color='m', label='prc')
pl.bar(x+0.25, prc2, width=0.25, color='k', label='prc1')
pl.legend(loc='upper right')
pl.ylabel("Price")
pl.xlabel("Week")
pl.show()
```

```python
# Horizontal bar

import matplotlib.pyplot as pl
week=[1,2,3,4]
prc=[45,56,34,23]
prc2=[48,58,53,89]
pl.barh(week,prc, height=[.4,.4,.3,.5], color=['r','k','b','m'])
pl.xlabel("Price")
pl.ylabel("Week")
pl.title('Data')
pl.show()
```

```python
# Piechart

import matplotlib.pyplot as pl
week=['A','B','C','D']
prc=[45,56,34,23]
prc2=[48,58,53,89]
pl.pie(prc, labels=week, autopct='%1.3f%%',colors=['g','orange','b','m'], explode=[0,0,0,.1])
pl.axis('equal')
pl.title('Data')
pl.show()
```

```python
# Histogram

import numpy as np
import matplotlib.pyplot as pl
ages=[1,1,3,5,6,6,7,8,8,9,9,10,10,11,12,12,14,14,15,15,
      20,23,24,25,25,25,27,29,30,30,35,36,37,37,39,
     40,43,45,48,48,50,51,52,52,55,57,59,60,63,65]
pl.hist(ages,bins=20, cumulative=False, histtype='step', orientation='horizontal')
pl.xlabel("Ages")
pl.ylabel("No. of Participants")
pl.title("Participant's Ages")
pl.show()
```

```python
import seaborn as sns

import matplotlib.pyplot as plt

titanic = sns.load_dataset("titanic")
# customize the bar plot

sns.barplot(x="class", y="fare", hue="sex", ci=None, palette="muted", data=titanic)

# add labels and title

plt.xlabel("Class")

plt.ylabel("Fare")

plt.title("Average Fare by Class and Gender on the Titanic")

# display the plot

plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import Lasso, Ridge
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import StandardScaler
from sklearn import datasets
```

```python
data = datasets.load_wine()
```

```python
X = data.data
```

```python
scaler = StandardScaler()
X = scaler.fit_transform(X)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, data.target, test_size = 0.2, random_state = 0)
```

```python
#implementing lasso regression
lasso = Lasso(alpha = 0.1)
lasso.fit(X_train, y_train)
y_pred = lasso.predict(X_test)
print("Predicted",y_pred)
print("Actual",y_test)
print('Mean Squared Error for Lasso Regression: ', mean_squared_error(y_test, y_pred))
```

```python
#implementing ridge regression
ridge = Ridge(alpha = 0.1)
ridge.fit(X_train, y_train)
y_pred = ridge.predict(X_test)
print("Predicted",y_pred)
print("Actual",y_test)
print('Mean Squared Error for Ridge Regression: ', mean_squared_error(y_test, y_pred))
```
